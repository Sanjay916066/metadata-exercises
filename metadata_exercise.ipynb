##### Metadata Creation Exercise

##### Goal
##### Practice creating metadata by applying DataCite and schema.org standards
to a public dataset.

##### Dataset Chosen
##### Titanic Passenger Survival Dataset

##### Source:
##### https://www.kaggle.com/c/titanic

```python
import json
```

##### Create DataCite XML Metadata
This code generates a DataCite metadata XML file.

```python
datacite_xml = """<?xml version="1.0" encoding="UTF-8"?>
<resource xmlns="http://datacite.org/schema/kernel-4"
xmlns:xsi="http://www.w3.org/2001/XMLSchema-instance"
xsi:schemaLocation="
http://datacite.org/schema/kernel-4
http://schema.datacite.org/meta/kernel-4.5/metadata.xsd">

<identifier identifierType="DOI">
10.0000/example.titanicdataset
</identifier>

<creators>
 <creator>
   <creatorName>Kaggle Contributors</creatorName>
   <affiliation>Kaggle</affiliation>
 </creator>
</creators>

<titles>
 <title>Titanic Passenger Survival Dataset</title>
</titles>

<publisher>Kaggle</publisher>

<publicationYear>2024</publicationYear>

<resourceType resourceTypeGeneral="Dataset">
Passenger Survival Dataset
</resourceType>

<subjects>
 <subject>Machine Learning</subject>
 <subject>Classification</subject>
 <subject>Survival Analysis</subject>
</subjects>

<dates>
 <date dateType="Created">2024-01-01</date>
</dates>

<formats>
 <format>CSV</format>
</formats>

<version>1.0</version>

<descriptions>
<description descriptionType="Abstract">
Titanic passenger survival dataset for machine learning.
</description>
</descriptions>

</resource>
"""

with open("titanic_datacite.xml","w") as f:
    f.write(datacite_xml)

print("DataCite XML created.")
```

#### Create schema.org JSON-LD Metadata
This code generates schema.org metadata in JSON-LD format.

```python
schema_data = {
 "@context":"https://schema.org",
 "@type":"Dataset",

 "name":"Titanic Passenger Survival Dataset",

 "description":
 "Passenger demographic and survival dataset for ML tasks.",

 "url":"https://www.kaggle.com/c/titanic",

 "creator":{
   "@type":"Organization",
   "name":"Kaggle Contributors"
 },

 "publisher":{
   "@type":"Organization",
   "name":"Kaggle"
 },

 "datePublished":"2024-01-01",

 "version":"1.0",

 "license":"https://www.kaggle.com/terms",

 "keywords":[
   "machine learning",
   "classification",
   "survival analysis"
 ],

 "distribution":{
   "@type":"DataDownload",
   "encodingFormat":"CSV",
   "contentUrl":"https://www.kaggle.com/c/titanic"
 },

 "variableMeasured":[
   "PassengerId",
   "Survived",
   "Age",
   "Sex",
   "Fare"
 ]
}

with open("titanic_schemaorg.json","w") as f:
    json.dump(schema_data,f,indent=4)

print("Schema.org JSON created.")
```

#### Create README File
Generate a README for GitHub submission.

```python
readme = """
# Metadata Exercise

Dataset:
Titanic Passenger Survival Dataset

Files:
- titanic_datacite.xml
- titanic_schemaorg.json

Metadata Standards:
- DataCite XML
- schema.org JSON-LD
"""

with open("README.md","w") as f:
    f.write(readme)

print("README created.")
```
